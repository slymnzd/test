```python
revenue_by_category = df.groupby('Product_Category')['Total_Amount'].sum()
```

```python
revenue_by_category.plot(kind='bar', figsize=(12, 8), color='grey')
```

```python
average_value_by_category = df.groupby('Product_Category')['Total_Amount'].mean()
```

```python
average_value_by_category.plot(kind='bar', figsize=(12, 8), color='green')
```

```python
avg_order_value = (
    df.groupby('Product_Category')['Total_Amount']
      .mean()
)

plt.figure(figsize=(10, 6))

avg_order_value.plot(kind='bar')

plt.title('Average Order Value by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Average Order Value')
plt.xticks(rotation=45)

plt.show()
```

```python
category_analysis = df.groupby('Product_Category').agg(
    Sales_Count=('Order_ID', 'count'),
    Total_Revenue=('Total_Amount', 'sum'),
    Average_Order_Value=('Total_Amount', 'mean')
).round(2)

category_analysis = category_analysis.sort_values(
    'Total_Revenue',
    ascending=False
)

category_analysis
```

```python
df.columns
```

```python
age_group_analysis = df.groupby('Age_Group').agg(
    Sales_Count=('Order_ID', 'count'),
    Total_Revenue=('Total_Amount', 'sum'),
    Average_Order_Value=('Total_Amount', 'mean')
)

age_group_analysis
```

```python
age_group_analysis.plot(kind='bar', figsize=(12, 8), subplots=True)
```

```python
import json

notebook_path = "e_commerce.ipynb"

with open(notebook_path, "r", encoding="utf-8") as f:
    notebook = json.load(f)

print("Cell sayı:", len(notebook["cells"]))

with open("e_commerce_content.txt", "w", encoding="utf-8") as f:
    for i, cell in enumerate(notebook["cells"], start=1):
        f.write("\n" + "=" * 70 + "\n")
        f.write(f"CELL {i} — {cell['cell_type'].upper()}\n")
        f.write("=" * 70 + "\n\n")

        source = "".join(cell.get("source", []))
        f.write(source)

        if cell["cell_type"] == "code":
            outputs = cell.get("outputs", [])

            if outputs:
                f.write("\n\n--- OUTPUT ---\n")

                for output in outputs:
                    if "text" in output:
                        f.write("".join(output["text"]))

                    elif "data" in output:
                        if "text/plain" in output["data"]:
                            f.write("".join(output["data"]["text/plain"]))

        f.write("\n")

print("e_commerce_content.txt yaradıldı.")
```
